```python
df_titanic = pd.read_csv('titanic.csv')

# Show first 5 rows
print("=== FIRST 5 ROWS ===")
print(df_titanic.head())
```

```python
print("=== DATA INFO ===")
print(df_titanic.info())

print("\n=== STATISTICS ===")
print(df_titanic.describe())
```

```python
df_clean = df_titanic.copy()

# 1. Fix missing Age values (fill with median age)
df_clean['Age'].fillna(df_clean['Age'].median(), inplace=True)

# 2. Fix missing Embarked values (fill with mode)
df_clean['Embarked'].fillna(df_clean['Embarked'].mode()[0], inplace=True)

# 3. Remove Cabin column (too many missing values)
df_clean.drop('Cabin', axis=1, inplace=True)

# 4. Remove duplicate rows if any
df_clean.drop_duplicates(inplace=True)

# Check if any missing values remain
print("=== MISSING VALUES AFTER CLEANING ===")
print(df_clean.isnull().sum())
```

```python
print("=== SURVIVAL RATE BY GENDER ===")
print(df_clean.groupby('Sex')['Survived'].mean())
print()

print("=== SURVIVAL RATE BY PASSENGER CLASS ===")
print(df_clean.groupby('Pclass')['Survived'].mean())
print()

print("=== AVERAGE AGE BY PASSENGER CLASS ===")
print(df_clean.groupby('Pclass')['Age'].mean())
```

```python
# Function to classify age
def age_group(age):
    if age < 18:
        return 'Child'
    else:
        return 'Adult'

# Apply the function to create new column
df_clean['AgeGroup'] = df_clean['Age'].apply(age_group)

# Survival rate by age group
print("=== SURVIVAL RATE BY AGE GROUP ===")
print(df_clean.groupby('AgeGroup')['Survived'].mean())
```

```python
# 1. Female survivors
female_survived = df_clean[(df_clean['Sex'] == 'female') & (df_clean['Survived'] == 1)]
print(f"=== FEMALE SURVIVORS: {len(female_survived)} passengers ===")
print(female_survived[['Name', 'Sex', 'Age', 'Pclass', 'Survived']].head())

# 2. Children who survived
children_survived = df_clean[(df_clean['Age'] < 18) & (df_clean['Survived'] == 1)]
print(f"\n=== CHILDREN WHO SURVIVED: {len(children_survived)} passengers ===")
print(children_survived[['Name', 'Age', 'Sex', 'Pclass', 'Survived']].head())

# 3. 1st class survivors
first_class_survived = df_clean[(df_clean['Pclass'] == 1) & (df_clean['Survived'] == 1)]
print(f"\n=== 1ST CLASS SURVIVORS: {len(first_class_survived)} passengers ===")
print(first_class_survived[['Name', 'Age', 'Sex', 'Pclass', 'Survived']].head())
```

```python
# Group by multiple columns
best_combination = df_clean.groupby(['Sex', 'Pclass', 'AgeGroup'])['Survived'].mean()
print("=== SURVIVAL RATE BY COMBINATION ===")
print(best_combination)

# Find the highest
highest_rate = best_combination.max()
highest_group = best_combination.idxmax()
print(f"\n=== HIGHEST SURVIVAL RATE: {highest_group} ===")
print(f"Rate: {highest_rate:.1%}")
```

```python
print("\n" + "="*50)
print("INSIGHTS FROM TITANIC DATA")
print("="*50)

print("\n1. Who was more likely to survive? (gender)")
print("   → Women had a much higher survival rate than men.")

print("\n2. Did class affect survival?")
print("   → Yes. 1st class passengers survived at the highest rate,")
print("     followed by 2nd class, then 3rd class.")

print("\n3. Were children prioritized?")
print("   → Yes. Children survived at a higher rate than adults.")

print("\n4. What combination had the highest survival rate?")
print(f"   → {highest_group} with {highest_rate:.1%} survival rate.")
```
